```python
import pandas as pd
from sklearn.datasets import fetch_openml

# Завантажуємо Ames Housing через офіційний API бібліотеки sklearn
try:
    housing = fetch_openml(name="house_prices", as_frame=True, parser='auto')
    df = housing.frame
    df.info() # Отримуємо стислий огляд нашого DataFrame (таблиці даних), бачимо кількість рядків, стовпців, їхні назви, кількість непустих значень та тип даних (наприклад, int64, float64, object) для кожного стовпця, а також загальне використання пам'яті
    print("Дані успішно завантажено!")
    print(f"Доступні колонки: {df.columns.tolist()[:10]}... (всього {len(df.columns)})")
except Exception as e:
    print(f"Помилка: {e}")

# Вивід перших 5 рядків
df.head()
```

```python
# 1. Вибір ключових колонок для проекту
# SalePrice - ціна, GrLivArea - площа, Neighborhood - район, 
# OverallQual - якість, YearBuilt - рік побудови
cols_to_keep = ['SalePrice', 'GrLivArea', 'Neighborhood', 'OverallQual', 'YearBuilt']
df_final = df[cols_to_keep].copy()

# 2. Очищення: видаляємо рядки з порожніми значеннями (якщо вони є)
df_final = df_final.dropna()

# 3. Перевірка результату
print("Очищений датасет готовий!")
print(f"Кількість записів: {len(df_final)}")
print(df_final.head())

# 4. Збереження файлу, який ми завантажимо на GitHub
df_final.to_csv('cleaned_real-estate-market-analysis.csv', index=False)
print("Файл 'cleaned_real-estate-market-analysis.csv' успішно створено!")
```
